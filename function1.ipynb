# 1. What is the difference between a function and a method in Python?

Function:
A function is a standalone block of code that performs a specific task. It is defined independently and can be called directly by its name. Functions operate on data passed to them as arguments and can return a value.

Method:
A method is a function that is associated with an object (an instance of a class). It is defined within the class and can access and modify the object's data (attributes). Methods are called on an object using dot notation (e.g., object.method()). The first parameter of a method is conventionally named self, which refers to the instance of the object.

Method vs. Function: Difference Between Methods and Function

Functions in Python	Methods in Python
Functions are outside a class	Methods are created inside a class
Functions are not linked to anything	Methods are linked with the classes they are created in
Functions can be executed just by calling with its name	To execute methods, we need to use either an object name or class name and a dot operator.
Functions can have zero parameters.	Methods should have a default parameter either self or cls to get the object’s or class’s address.
Functions can not access or modify class attributes	Methods can access and modify class attributes
Functions are independent of classes	Methods are dependent on classes
By looking at the above differences, we can simply say that all methods are functions

#3. What are the different ways to define and call a function in Python?

In Python, functions can be defined and called in several ways:
1. Standard Function Definition:-
  This is the most common way to define a function, using the def keyword, followed by the function name, parentheses for arguments, and a colon.

```python
def greet(name):
    return "Hello, " + name

# Calling the function
message = greet("World")
print(message)
```

2. Lambda Functions (Anonymous Functions)

Lambda functions are small, anonymous functions defined using the lambda keyword. They can take any number of arguments but can only have one expression.

```python
add = lambda x, y: x + y

# Calling the lambda function
result = add(5, 3)
print(result)
```

3. Nested Functions

A function can be defined inside another function. The inner function can access the outer function's variables.

```python
def outer_function(text):
    def inner_function():
        return "Inner: " + text
    return inner_function()

# Calling the outer function
message = outer_function("Hello")
print(message)
```

4. Functions with Default Arguments

ou can provide default values for function arguments, which are used if the caller doesn't provide a value for that argument.

```python
def power(base, exponent=2):
    return base ** exponent

# Calling with both arguments
result1 = power(3, 3)
print(result1)

# Calling with only the base (using default exponent)
result2 = power(5)
print(result2)
```

5. Functions with Variable Number of Arguments

*args allows a function to accept any number of positional arguments.
**kwargs allows a function to accept any number of keyword arguments (named arguments)

```python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Calling with *args
product = multiply(2, 3, 4)
print(product)

# Calling with **kwargs
print_info(name="Alice", age=30, city="New York")
```

#5. What are iterators in Python and how do they differ from iterables?

An iterator is an object that contains a countable number of values. An iterator is an object that can be iterated upon, meaning that you can traverse through all the values. Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods __iter__() and __next__() .

Difference Between Iterable and Iterator in Python

The main differences between iterables and iterators are:
Iterables can be looped over multiple times, while iterators can only be traversed once.
Iterables can be indexed, while iterators cannot.
Iterables can be copied, while iterators cannot.
Iterables can be used in for loops and other constructs that require iteration, while iterators can only be used with the next() function.
Iterables store all their elements in memory, while iterators generate elements on demand, which can be more memory-efficient for large datasets.

```python
# Example of an iterable (list)
my_list = [1, 2, 3]

# Create an iterator from the iterable
my_iterator = iter(my_list)

# Iterate through the iterator
print(next(my_iterator)) # Output: 1
print(next(my_iterator)) # Output: 2
print(next(my_iterator)) # Output: 3

# The iterator is exhausted, so it raises StopIteration
try:
    print(next(my_iterator))
except StopIteration:
    print("Iterator is exhausted")
```

#6. Explain the concept of generators in Python and how they are defined.

A Python generator function allows you to declare a function that behaves like an iterator, providing a faster and easier way to create iterators. They can be used on an abstract container of data to turn it into an iterable object like lists, dictionaries and strings.

Generator Function in Python

A generator function is a special type of function that returns an iterator object. Instead of using return to send back a single value, generator functions use yield to produce a series of results over time. This allows the function to generate values and pause its execution after each yield, maintaining its state between iterations.

```python
def fun(max):
    cnt = 1
    while cnt <= max:
        yield cnt
        cnt += 1

ctr = fun(5)
for n in ctr:
    print(n)
```

#7. What are the advantages of using generators over regular functions?

Generators offer several advantages over regular functions, primarily stemming from their ability to produce values on-demand and manage state efficiently. This makes them memory-efficient, ideal for handling large or infinite sequences, and suitable for tasks requiring lazy evaluation.

Memory Efficiency:
Unlike regular functions that return the entire result at once, generators yield values one by one as needed, reducing memory consumption, especially when dealing with large datasets.

Lazy Evaluation: Generators compute values only when they are requested, which can improve performance, particularly in scenarios where not all values are required.

Concise Code:Generators often provide a cleaner and more readable way to create iterators compared to traditional loops.

State Management:generators maintain their state between calls, allowing you to resume execution from the last yielded point.

Infinite Sequences:Generators can be used to create infinite sequences of values, unlike regular functions that need to return a complete result.

Cooperative Multitasking:Generators can be paused and resumed, facilitating cooperative multitasking where different parts of the code can share the processing load.

#8. What is a lambda function in Python and when is it typically used?

In Python, a lambda function is a concise, anonymous function defined using the lambda keyword. It's typically used for creating small, one-time-use functions, especially when you need a function as an argument to another function (like map, filter, or sorted).

Key aspects of lambda functions:

Anonymous:They don't have a name like regular functions defined with def.

Concise:They are limited to a single expression, making them suitable for simple operations.

Used with higher-order functions:They are often passed as arguments to functions that take other functions as input.

One-time use:They are designed for short, temporary functions where a full function definition might be overkill.

When to use lambda functions:

With higher-order functions: When using map, filter, or sorted where you need a simple function to apply or test each element.
As arguments to other functions: When you need a small, anonymous function as an argument.
For simple, single-line expressions: When the operation you need to perform is simple and can be expressed in a single line.
Example:

```python
# Using a lambda function with map() to square each number in a list
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))  # The lambda function squares the input 'x'
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

#9. Explain the purpose and usage of the `map()` function in Python.

The map() function in Python serves to apply a given function to each item of an iterable (like a list or tuple) and returns an iterator that yields the results. This eliminates the need for explicit loops, making code more concise and readable.

map() takes two arguments:A function: This is the function that will be applied to each item in the iterable.
An iterable: This is the sequence of items (e.g., list, tuple) that will be processed

The basic syntax is:

```python
map(function, iterable, ...)
```

The function can be a built-in function, a user-defined function, or a lambda function. If multiple iterables are passed, the function must accept that many arguments, and the map() function will stop when the shortest iterable is exhausted.

```python
def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5]
squared_numbers = map(square, numbers)

print(list(squared_numbers)) # Output: [1, 4, 9, 16, 25]
```

the square function is applied to each element of the numbers list. map() returns a map object, which is then converted to a list for printing. Lambda functions can also be used directly within map() for more concise operations:

```python
numbers = [1, 2, 3, 4, 5]
squared_numbers = map(lambda x: x * x, numbers)
print(list(squared_numbers)) # Output: [1, 4, 9, 16, 25]
```

#10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

map(), filter(), and reduce() are built-in functions in Python that operate on iterables (like lists, tuples, etc.) and are often used in functional programming. They each serve distinct purposes:

map():Applies a given function to each item in an iterable and returns a new iterator with the results. It transforms each element individually.

filter():Creates a new iterator containing only the items from the original iterable for which a given function returns True. It selects elements based on a condition.

reduce():Applies a function cumulatively to the items of an iterable, from left to right, to reduce the iterable to a single value. It aggregates elements

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map(): squares each number
squared_numbers = list(map(lambda x: x**2, numbers))  # Output: [1, 4, 9, 16, 25]

# filter(): selects even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))  # Output: [2, 4]

# reduce(): calculates the product of all numbers
product = reduce(lambda x, y: x * y, numbers)  # Output: 120
```

#4. Write a Python function that checks if a given number is prime or not from 1 to 200.




Python function that checks whether a given number from 1 to 200 is prime:

```python
def is_prime(n):
    if n < 2 or n > 200:
        return False  # not prime or out of allowed range
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

example:

```python
print(is_prime(7))    # Output: True
print(is_prime(10))   # Output: False
print(is_prime(201))  # Output: False (outside range)
```

#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.


an implementation of an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms:

```python
class FibonacciIterator:
    def __init__(self, num_terms):
        self.num_terms = num_terms
        self.current, self.next_term = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.num_terms:
            raise StopIteration
        else:
            current_term = self.current
            self.current, self.next_term = self.next_term, self.current + self.next_term
            self.count += 1
            return current_term
```

example:

```python
fib = FibonacciIterator(10)

for term in fib:
    print(term)
```
